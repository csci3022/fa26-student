## 🔴 Question 4: Shannon Entropy and Efficient Communication (9 points)


In 1948, Claude Shannon developed a mathematical theory for a basic communication problem:

> **If a source produces some symbols much more often than others, how efficiently can we represent a long message using bits without losing any information?**

Suppose a communication source produces one symbol at a time. Let $X$ be the discrete random variable representing the **next symbol produced by the source**.

Thus, $X$ can take one of four possible values:

$$
X \in \{A,B,C,D\}.
$$

The probability mass function (PMF) of $X$ is:

| $x$ | $P(X=x)$ |
| --- | ---: |
| $A$ | $1/2$ |
| $B$ | $1/4$ |
| $C$ | $1/8$ |
| $D$ | $1/8$ |

Assume that successive symbols are generated **independently according to this same PMF**. For example, $P(X=A)=1/2$ means that in a long sequence produced by this source, we expect about half of the symbols to be $A$.

We will represent the sequence of symbols using **bits**. A **bit** (short for **binary digit**) is one `0` or `1`.

### Start with a simple encoding that ignores the probabilities

The most straightforward approach is to give every symbol a codeword of the same length. Since there are four possible symbols, we can use the four two-bit strings

$$
A=00,\qquad B=01,\qquad C=10,\qquad D=11.
$$

This is a **fixed-length code**: every symbol uses exactly 2 bits. It works perfectly well, but it completely ignores the fact that $A$ occurs much more often than $C$ or $D$.

For example, a message containing 1,000 symbols would always require 2,000 bits with this code, no matter how often each symbol occurs.

That raises the question Shannon wanted to formalize: **Can we take advantage of the probabilities to use fewer bits on average? And what is the theoretical limit on how much compression is possible?**

## 🔴 Question 4b — measuring the information in the source (2 points)

To answer the compression question, Shannon associated an amount of **self-information** with each possible outcome. If outcome $x$ has probability $P(X=x)$, its self-information is

$$
\boxed{I(x)=-\log_2(P(X=x))}
$$

and the units are **bits**.

### Why does this formula involve $\log_2$, and why are the units bits?

Recall that a **bit** is a binary digit (`0` or `1`). The important fact is:

> **With $k$ bits, there are $2^k$ different possible bit strings.**

For example, with 3 bits there are $2^3=8$ possible strings:

$$
000,\ 001,\ 010,\ 011,\ 100,\ 101,\ 110,\ 111.
$$

So if there are 8 equally likely possibilities, 3 bits are enough to identify exactly which one occurred. Similarly,

$$
2\text{ possibilities}\longleftrightarrow 1\text{ bit},\qquad
4\text{ possibilities}\longleftrightarrow 2\text{ bits},\qquad
8\text{ possibilities}\longleftrightarrow 3\text{ bits}.
$$

This gives the connection to probability. An event with probability $1/8$ has the same probability as one particular outcome among 8 equally likely possibilities. Distinguishing among 8 equally likely possibilities requires 3 bits because $2^3=8$. Thus,

$$
P(X=x)=\frac18 \quad\Rightarrow\quad I(x)=3\text{ bits}.
$$

Likewise,

$$
P(X=x)=\frac12 \quad\Rightarrow\quad I(x)=1\text{ bit},
$$

and

$$
P(X=x)=\frac14 \quad\Rightarrow\quad I(x)=2\text{ bits}.
$$

More generally, suppose

$$
P(X=x)=\frac{1}{2^k}.
$$

Then

$$
\frac{1}{P(X=x)}=2^k.
$$

Taking $\log_2$ answers the question:

> **What power of 2 is the reciprocal of this probability?**

Thus,

$$
k=\log_2\left(\frac{1}{P(X=x)}\right)
=-\log_2(P(X=x)).
$$

That is why the self-information formula is

$$
I(x)=-\log_2(P(X=x)).
$$

The **base 2** is used because we are measuring information using binary digits, so the units are **bits**. The **negative sign** appears because probabilities are between 0 and 1, so $\log_2(P(X=x))$ is zero or negative. Multiplying by $-1$ makes the information nonnegative. It also gives the behavior we want: **rarer outcomes have larger self-information because learning that a rare event occurred is more informative (or surprising).**

Importantly, $I(x)$ is an **amount of information**, not necessarily the literal number of `0`s and `1`s in the codeword used for a single symbol. Individual binary codewords must have integer lengths, while $-\log_2(p)$ can be fractional.

### Shannon entropy

Because $X$ is a random variable, $I(X)$ is also a **random variable**: its value depends on which symbol $X$ takes. Shannon defined the **entropy** of the source to be the expected value of this random variable:

$$
\boxed{H(X)=E[I(X)]
=\sum_x P(X=x)I(x)
=-\sum_x P(X=x)\log_2(P(X=x))}.
$$

The self-information $I(x)$ is measured in **bits**, so $H(X)=E[I(X)]$ is also measured in **bits**. Since $X$ represents one symbol produced by the source, we often interpret entropy as the **average number of bits of information per symbol**. For this reason, entropy is also commonly described in units of **bits per symbol**.

> It turns out that Shannon entropy gives a theoretical benchmark for lossless compression. For long messages, efficient lossless encodings can make the average number of bits used per symbol approach $H(X)$, but cannot systematically beat this limit. This result is known as **Shannon's source coding theorem**. We will not prove it here, but in this question we will see the idea in action.

### Entropy versus code length

In our example, the variable-length code has codeword lengths

$$
1,\ 2,\ 3,\ 3,
$$

which happen to be exactly the self-information values

$$
-\log_2(1/2)=1,\qquad
-\log_2(1/4)=2,\qquad
-\log_2(1/8)=3.
$$

Therefore, in this particular example,

$$
E[L]=H(X).
$$

This exact equality does **not** occur for every probability distribution. Codewords must contain a whole number of bits, while self-information values such as $-\log_2(p)$ do not have to be whole numbers.

Also notice the distinction between the two quantities:

- $H(X)$ is determined by the **probability distribution of the source**.
- $L$ is determined by the **particular encoding we choose**, and $E[L]$ is its average codeword length.

It turns out that for any **uniquely decodable binary code** for these symbols,

$$
H(X)\leq E[L].
$$

Thus, Shannon entropy gives a theoretical lower bound on the average number of bits needed per symbol. A code that assigns one codeword to each individual symbol may not achieve this lower bound exactly. However, by encoding long blocks of independent source symbols together, efficient lossless codes can make the average number of bits per symbol approach $H(X)$.

### Connection to generative AI

These same ideas appear in modern generative AI. A large language model predicts a probability distribution over the next token (a word or piece of a word). If the model assigns substantial probability to many different possible next tokens, the distribution has higher entropy, meaning there is more uncertainty about what comes next. If the model places most of its probability on just a few tokens, the distribution has lower entropy.

A closely related quantity, **cross-entropy**, is used when training language models to measure how well the model's predicted probability distribution matches the tokens that actually occur. We will return to cross-entropy later in this course.

## 🔴 Question 5: Binomial Probabilities and Sampling (6 points)

**Sampling With Replacement**


Consider a population in which a proportion $p$ of individuals are called "successes" (or 1, if you prefer) and the remaining proportion are rudely called "failures" (or 0).

We can sample from this population **at random with replacement**.  At random with replacement means we randomly choose one individual from the population and record whether or not they are a "success" or a "failure."  We then return that individual to the population before sampling again.  

(The reason why the individual is returned to the population before the next individual is drawn is to make sure the probability of selecting any particular individual remains the same in future draws).


If you draw a sample of size $n$ (where $n$ is some positive integer) at random with replacement from the population, then  probability of drawing $k$ successes and $n-k$ failures is $\binom{n}{k}p^k(1-p)^{n-k}$, for $0 \leq k \leq n$. To reduce writing, we will shorten "$k$ successes and $n-k$ failures" to "$k$ successes". In other words, "$k$ successes" means "exactly $k$ successes".

Since this random variable shows up in many applications it has an official mathematical name:  It is called a **binomial random variable** and its probability distribution is called a 
 **binomial distribution**.

To formalize here's how we write this in mathematical notation: 

 - We write $S\sim Bin(n, p)$ to represent that **the random variable $S$ is distributed as a Binomial Distribution with parameters $n$ and $p$.**  

 - The PMF (probability mass function) for $S$ is 
$P(S = k) = \binom{n}{k}p^k(1-p)^{n-k}$.

**Calculating binomial coefficients in Python**

Factorials and the *binomial coefficients* $\binom{n}{k} = \frac{n!}{k!(n-k)!}$ get large very quickly as $n$ gets large. One way to compute them is to use the `SciPy` module `special`. `SciPy` is a collection of Python-based software for math, probability, statistics, science, and engineering.

```python
from scipy import special
```

Below are some examples of `special.factorial`:

```python
special.factorial(5)

special.factorial(range(1, 6))
```

Traditionally, subsets of $k$ individuals out of a population of $n$ individuals are called *combinations*, and so `special.comb(n, k)` evaluates to $\binom{n}{k}$. 

Note that we will always use the term *subsets* to mean un-ordered sets. We will use *permutations* in situations where we need to keep track of the order in which the elements appear.


Look at the code and output below carefully (including types) to understand how these functions work.


https://docs.scipy.org/doc/scipy/reference/generated/scipy.special.comb.html

Consider (once again) a sample of size $n$ drawn at random with replacement from a population in which a proportion $p$ of the individuals are called successes.

Let $S$ be the random variable that denotes the number of successes in our sample. (As stated above, $S$ follows the **binomial distribution**.) Then, the probability that the number of successes in our sample is **at most** $s$ (where $0 \leq s \leq n$) is

$$P(S \leq s) = P(S = 0) + P(S = 1) + ... + P(S = s) = \sum_{k=0}^s \binom{n}{k}p^k(1-p)^{n-k}$$

We obtain this by summing the probability that the number of successes is exactly $k$, for each value of $k = 0, 1, 2, ..., s$.

We have a special name for the function $F(s) = P(S \leq s)$, for any given random variable $S$.   

It is called the **Cumulative Distribution Function (CDF)** of $S$.
