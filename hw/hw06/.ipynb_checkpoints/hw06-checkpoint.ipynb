### Optional Practice 3: Empirical vs. theoretical Poisson distributions
Compare a simulated sample from $Poi(2.79)$ with the theoretical PMF. Try more than one sample size and notice how the empirical distribution changes as the sample size increases.


```python
# Optional practice: change sample_size and rerun this cell to see what happens
sample_size = 100
sample = stats.poisson.rvs(mu=2.79, size=sample_size)
bins = np.arange(-0.5, 15, 1)

# Theoretical probabilities
max_x = 15
X = np.arange(max_x + 1)
pmf = stats.poisson.pmf(X, mu=2.79)

fig, ax = plt.subplots(nrows=2, ncols=1)
plt.subplots_adjust(hspace=0.5)

# Histogram of simulated values
ax[0].hist(sample, edgecolor='white', density=True, bins=bins)
ax[0].set_axisbelow(True)
ax[0].grid(alpha=0.25)
ax[0].set_xlim([0, 15])
ax[0].set_title(f"Simulated Poisson Distribution: lambda = 2.79, n = {sample_size}")

# Theoretical probabilities
ax[1].bar(X, pmf, edgecolor='white', width=1)
ax[1].set_axisbelow(True)
ax[1].grid(alpha=0.25)
ax[1].set_xlim([0, 15])
ax[1].set_title("Theoretical Poisson Distribution, lambda = 2.79")

plt.show()
```

In computer science and machine learning, we often use data to estimate an unknown probability. For example, we might want to estimate the probability that a user clicks a link, that a system succeeds, or that a classifier makes a correct prediction.

In this problem, you'll see how **Bayes' Rule** can be used to update what we know about an unknown probability after observing some data. Along the way, we will encounter a new continuous distribution called the **Beta distribution**.





### A note about PDF notation

So far, we have usually referred to a PDF simply as $f(x)$. When a problem contains more than one random variable, it is helpful to indicate **which random variable** the PDF belongs to. We will use a subscript for this:

$$f_X(x) = \text{the PDF of the random variable }X.$$ 

For example, in this problem $f_\Theta(\theta)$ means **the PDF of the random variable $\Theta$**, evaluated at the possible value $\theta$. The subscript does not change what a PDF is; it simply makes the notation clearer when several random variables appear in the same problem.

### Bayes' Rule for a continuous parameter

You have previously used Bayes' Rule for events:

$$
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}.
$$

Recall that the denominator, $P(B)$, can be found using the **Law of Total Probability**.

Suppose $A_1,A_2,\ldots,A_k$ represent all of the possible, mutually exclusive cases. Then

$$
P(B)=\sum_{i=1}^k P(B\mid A_i)P(A_i).
$$

Substituting this into Bayes' Rule gives

$$
\boxed{
P(A_j\mid B)
=
\frac{P(B\mid A_j)P(A_j)}
{\displaystyle\sum_{i=1}^k P(B\mid A_i)P(A_i)}
}
$$

So, when the possibilities are **discrete**, the denominator accounts for all possible cases by **adding**:

$$
\sum_{\text{all possible cases}}
(\text{likelihood})(\text{prior probability}).
$$

---

The same idea applies when the unknown quantity is **continuous**.

In our website example, $\Theta$ can take **any value between 0 and 1**. Because there are infinitely many possible values of $\theta$, instead of adding with a sum, we account for all possible values using an **integral**:

$$
\boxed{
f_{\Theta\mid X=3}(\theta)
=
\frac{
P(X=3\mid\Theta=\theta)\,f_\Theta(\theta)
}{
\displaystyle
\int_0^1
P(X=3\mid\Theta=\theta)\,f_\Theta(\theta)\,d\theta
}
}
$$

Notice the parallel:

$$
\begin{array}{c|c}
\textbf{Discrete Bayes} & \textbf{Continuous Bayes}\\[4pt]
P(A_i) & f_\Theta(\theta)\\[4pt]
P(B\mid A_i) & P(X=3\mid\Theta=\theta)\\[4pt]
\displaystyle\sum_i & \displaystyle\int
\end{array}
$$

For our problem:

- $f_\Theta(\theta)$ is the **prior**: what we believed about $\Theta$ before seeing the data.
- $P(X=3\mid\Theta=\theta)$ is the **likelihood**: the probability of observing our data if the click probability were $\theta$.
- $f_{\Theta\mid X=3}(\theta)$ is the **posterior**: our updated PDF for $\Theta$ after observing the data.

The denominator accounts for **all possible values of the unknown click probability**, just as the sum in the Law of Total Probability accounts for all possible cases in the discrete version of Bayes' Rule.

The denominator is also called the **normalizing constant** because it makes the posterior a valid PDF that integrates to 1.

### 🔴 Question 7e (1 point)

The posterior PDF you found in Question 7d is an example of a **Beta distribution**.

A continuous random variable $Y$ has a Beta distribution with parameters $\alpha$ and $\beta$ when its PDF has the form

$$
f_Y(y)=\frac{1}{B(\alpha,\beta)}
y^{\alpha-1}(1-y)^{\beta-1},
\qquad 0\leq y\leq1.
$$

The quantity $B(\alpha,\beta)$ is a constant that makes the total area under the PDF equal to 1. You do **not** need to calculate this constant.

To identify the parameters $\alpha$ and $\beta$, we only need to look at the **exponents**:

- the exponent on $y$ is $\alpha-1$
- the exponent on $(1-y)$ is $\beta-1$

For example, if a Beta PDF contained the expression

$$
y^4(1-y)^2,
$$

then we would match the exponents:

$$
\alpha-1=4
\qquad\text{and}\qquad
\beta-1=2,
$$

so this would have $\alpha=5$ and $\beta=3$.

### What just happened?

Before observing the data, our **prior** was $Uniform(0,1)$, so every value of $\theta$ between 0 and 1 had the same density.

After observing **3 clicks out of 4 visitors**, Bayes' Rule updated that prior using the data. Values of $\theta$ that better explain what we observed now have a higher **posterior density**. In this example, our posterior distribution is

$$
\Theta\mid X=3 \sim Beta(4,2).
$$

**One useful connection:** $Uni(0,1)$ is the same distribution as $Beta(1,1)$. More generally, when we start with a **Beta prior** for an unknown probability and observe **Binomial data**, the posterior is also a Beta distribution.

This is one reason the Beta distribution is especially useful for learning about unknown probabilities from binary data.

### Visualization: Prior vs. Posterior

Run the provided code below to compare the $Uni(0,1)$ **prior** with the $Beta(4,2)$ **posterior**. There is nothing to submit for this visualization.

As you look at the graph, compare how the density is distributed **before and after observing 3 clicks out of 4 visitors**.

Notice that the posterior places more density on larger values of $\theta$, since those values are more consistent with the data we observed. However, the posterior still spreads density over a range of possible values, representing our remaining uncertainty about the true value of $\Theta$.

```python
theta = np.linspace(0, 1, 500)

prior = stats.uniform.pdf(theta, loc=0, scale=1)
posterior_density = stats.beta.pdf(theta, a=4, b=2)

plt.plot(theta, prior, label='Prior: Uniform(0,1)')
plt.plot(theta, posterior_density, label='Posterior: Beta(4,2)')
plt.xlabel(r'$\theta$')
plt.ylabel('Probability density')
plt.title('Prior and posterior distributions for the click probability')
plt.legend()
plt.show()
```
